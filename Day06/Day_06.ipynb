# Part 1

```python
file_name = "input.txt"
# file_name = "test.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")
operators = data[-1].split()

data: list[list[str]] = [i.split() for i in data[:-1]]

res = 0 
for i in range(len(data[0])):
    col_res = ""
    op = operators[i]
    for j in range(len(data)):
        col_res += data[j][i] + op
    res += int(eval(col_res[:-1]))
        
res
```
