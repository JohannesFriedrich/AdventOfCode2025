# Part 1

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n\n")

ranges = [range(int(a), int(b)+1)  for i in data[0].split("\n") for a,b in [i.split("-")]]

IDs = [int(i) for i in data[1].split("\n")]

res = 0 
for ID in IDs:
    found = False
    for r in ranges:
        if ID in r:
            res += 1
            found = True
            break
    if found:
        continue          

res
```

# Part 2

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n\n")

ranges = [(int(a),int(b)) for i in data[0].split("\n") for a,b in [i.split("-")]]
ranges = sorted(ranges)

res = 0
max_index = 0
for start, end in ranges:
    if end >= max_index:
        res += end - max(start, max_index) + 1
        max_index = end + 1

res
```
