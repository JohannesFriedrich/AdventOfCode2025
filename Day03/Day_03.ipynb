# Part 1

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

res = 0

for bank in data:
    nums = [int(i) for i in bank]
    n_max_1 = max(nums[:len(nums)-1])
    max_pos = nums.index(n_max_1)
    n_max_2 = max(nums[max_pos+1:])
    res+= 10*n_max_1 + n_max_2

res
```

# Part 2

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")
res = 0
def get_max(nums, index, max_pos):
    n_max = max(nums[max_pos+1:len(nums)-index+1])
    new_max_pos = nums.index(n_max, max_pos+1)
    return [n_max, new_max_pos]

for bank in data:
    nums = [int(i) for i in bank]
    bank_res = ""
    max_pos = -1
    for i in range(12,0,-1):
        n_max, max_pos = get_max(nums, i, max_pos)
        bank_res += str(n_max)
    res += int(bank_res)

res
```
