# Part 1

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split(",")
res = 0

for IDs in data:
    first, last = IDs.split("-")
    for ID in range(int(first), int(last)+1):
        l = len(str(ID))
 
        if l % 2 == 0:
            left = str(ID)[:int(l/2)]
            right = str(ID)[int(l/2):]
            if left == right:
                res += ID


res
```

# Part 2

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split(",")
res = 0

def check_ID(ID):
    ID_s = str(ID)
    l= len(ID_s)
    max_part = int(l/2)
    for n in range(max_part+1, 0, -1):
        splits = [ID_s[i:i+n] for i in range(0, len(ID_s), n)]
        if splits.count(splits[0]) == len(splits) and len(splits) >= 2:
            return True
    return False
    

for IDs in data:
    first, last = IDs.split("-")
    for ID in range(int(first), int(last)+1):
        if check_ID(ID):
            res += ID
            


res
```
