# Part 1

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

cur_pos = 50
LEN = 100
res = 0

for step in data:
    dir = step[0]
    amount = int(step[1:]) if dir == "R" else -int(step[1:]) 
    cur_pos = (cur_pos + amount) % LEN
    if cur_pos == 0:
        res += 1
res
```

# Part 2

```python
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

cur_pos = 50
LEN = 100
res = 0

for step in data:
    dir = step[0]
    amount = int(step[1:])

    if dir == 'L':
        cur_pos = (100 - cur_pos) % 100

    cur_pos += amount
    res += cur_pos // 100
    cur_pos %= 100

    if dir == 'L':
        cur_pos = (100 - cur_pos) % 100

res
```
