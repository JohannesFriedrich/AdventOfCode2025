```python
from math import prod
file_name = "input.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

data = [[int(i) for i in line.split(",")] for line in data]

dist_map = {}
for i1,d1 in enumerate(data):
    for i2,d2 in enumerate(data):
        if i1 < i2:
            dist = (d1[0]-d2[0])**2 + (d1[1]-d2[1])**2 + (d1[2]-d2[2])**2
            dist_map[str(i1)+"-"+str(i2)] = dist

dist_map = {k: v for k, v in sorted(dist_map.items(), key=lambda item: item[1])}
        
circuits_l = []

i = 0
for con in dist_map:
    if i >=1000:
        break
    jb1, jb2 = con.split("-")
    updated = False
    jb1_found = False
    jb2_found = False
    # checks
    for idx, circuits in enumerate(circuits_l):
        if jb1 in circuits:
            jb1_found = True
            jb1_idx = idx
        if jb2 in circuits:
            jb2_found = True
            jb2_idx = idx
        if jb1_found and jb2_found:
            break

    if jb1_found and jb2_found and jb1_idx != jb2_idx:
        circuits_l[jb1_idx].update(circuits_l[jb2_idx])
        del circuits_l[jb2_idx]
    elif jb1_found:
        circuits_l[jb1_idx].update({jb1, jb2})
    elif jb2_found:
        circuits_l[jb2_idx].update({jb1, jb2})
        updated = True
    else:
        circuits_l.append({jb1, jb2})
    i += 1

prod([len(i) for i in sorted(circuits_l, key=len, reverse=True)[:3]])
```
